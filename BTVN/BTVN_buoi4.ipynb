```python
# Hút bụi 

import random

ROWS = 4
COLS = 4

# Môi trường
def init_environment():
    grid = []
    for r in range(ROWS):
        row = []
        for c in range(COLS):
            row.append(0)
        grid.append(row)
    
    positions = []
    for r in range(ROWS):
        for c in range(COLS):
            positions.append((r, c))
            
    obstacles = random.sample(positions, 2)
    for r, c in obstacles:
        grid[r][c] = -1
        
    empty_positions = []
    for p in positions:
        if p not in obstacles:
            empty_positions.append(p)
            
    dirts = random.sample(empty_positions, 2)
    for r, c in dirts:
        grid[r][c] = 1
        
    agent_pos = random.choice(empty_positions)
    
    return grid, agent_pos

def get_valid_moves(r, c, grid):
    moves = []
    if r > 0 and grid[r-1][c] != -1:
        moves.append(('U', r-1, c))
    if r < ROWS - 1 and grid[r+1][c] != -1:
        moves.append(('D', r+1, c))
    if c > 0 and grid[r][c-1] != -1:
        moves.append(('L', r, c-1))
    if c < COLS - 1 and grid[r][c+1] != -1:
        moves.append(('R', r, c+1))
    return moves

def choose_action(r, c, grid, memory):
     # Nếu bộ nhớ cho biết ô hiện tại có rác (1), hành động là Hút (S)
    if memory[r][c] == 1:
        return 'S', (r, c)
    
    valid_moves = get_valid_moves(r, c, grid)
    
    # Ưu tiên: Nếu thấy ô hàng xóm nào có rác (đã nhớ), đi tới ô đó ngay
    for action, nr, nc in valid_moves:
        if memory[nr][nc] == 1:
            return action, (nr, nc)

     # Nếu không thấy rác, tìm các ô chưa bao giờ đi qua (None) để khám phá       
    unknown_moves = []
    for a, nr, nc in valid_moves:
        if memory[nr][nc] == None:
            unknown_moves.append((a, nr, nc))

     # Nếu có ô chưa biết, chọn ngẫu nhiên một ô để đi vào        
    if len(unknown_moves) > 0:
        a, nr, nc = random.choice(unknown_moves)
        return a, (nr, nc)

    # Nếu không có đường nào đi được, Robot bị kẹt (K)    
    if len(valid_moves) == 0:
        return 'K', (r, c)

    # Cuối cùng, nếu ô nào cũng biết rồi thì chọn đại một hướng đi hợp lệ    
    a, nr, nc = random.choice(valid_moves)
    return a, (nr, nc)

def print_grid(grid, r, c):
    header = "   "
    for j in range(COLS):
        header += " C" + str(j) + " "
    print(header)
    
    for i in range(ROWS):
        row_str = "R" + str(i) + " |"
        for j in range(COLS):
            if i == r and j == c:
                row_str += "[R]|"
            elif grid[i][j] == -1:
                row_str += " X |"
            else:
                row_str += " " + str(grid[i][j]) + " |"
        print(row_str)
    print()

# main
grid, agent_pos = init_environment()

# khởi tạo bộ nhớ (None)
memory = []
for r in range(ROWS):
    row = []
    for c in range(COLS):
        row.append(None)
    memory.append(row)

print("=== BẮT ĐẦU ===")
print_grid(grid, agent_pos[0], agent_pos[1])

for step in range(50):
    r = agent_pos[0]
    c = agent_pos[1]
    
    # Quan sát ô hiện tại & nhớ trạng thái (sạch/rác)
    memory[r][c] = grid[r][c]
    
    # chọn hàng dựa trên vị trí, bộ nhớ
    action, next_pos = choose_action(r, c, grid, memory)
    
    # hút
    if action == 'S':
        print("Bước " + str(step + 1) + ": HÚT SẠCH tại (" + str(r) + ", " + str(c) + ")")
        grid[r][c] = 0
        memory[r][c] = 0
    # kẹt
    elif action == 'K':
        print("Bước " + str(step + 1) + ": BỊ KẸT tại (" + str(r) + ", " + str(c) + ")")
    # di chuyển
    else:
        print("Bước " + str(step + 1) + ": DI CHUYỂN " + action)
        agent_pos = next_pos
        
    print_grid(grid, agent_pos[0], agent_pos[1])
    
    # kiểm tra
    total_dirt = 0
    for i in range(ROWS):
        for j in range(COLS):
            if grid[i][j] == 1:
                total_dirt += 1
                
    if total_dirt == 0:
        print("DONE!")
        break
```

```python
# 8-puzzles BFS

board = [[1, 2, 3], 
         [5, 7, 6], 
         [4, 8, 0]]

GOAL  = [[1, 2, 3], 
        [4, 5, 6], 
        [7, 8, 0]]

def in_bang(board, title=""):
    if title:
        print(title)
    for row in board:
        dong = ""
        for x in row:
            if x == 0:
                dong = dong + "_ "
            else:
                dong = dong + str(x) + " "
        print(dong)
    print()

def tim_o_trong(board):
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c

def di_chuyen(board, action):
    new_board = []
    for row in board:
        new_board.append(row[:])

    r, c = tim_o_trong(new_board)

    if action == 'U':
        nr, nc = r - 1, c
    elif action == 'D':
        nr, nc = r + 1, c
    elif action == 'L':
        nr, nc = r, c - 1
    else:
        nr, nc = r, c + 1

    temp = new_board[r][c]
    new_board[r][c] = new_board[nr][nc]
    new_board[nr][nc] = temp

    return new_board

def lay_hanh_dong(board):
    r, c = tim_o_trong(board)
    tat_ca = []

    if r > 0: tat_ca.append('U')
    if r < 2: tat_ca.append('D')
    if c > 0: tat_ca.append('L')
    if c < 2: tat_ca.append('R')

    return tat_ca

# BFS
def breadth_first_search(initial_state):
    node = {"state": initial_state, "parent": None, "action": None}
    
    if initial_state == GOAL:
        return node
        
    frontier = [node]
    reached = []
    
    while len(frontier) > 0:
        node = frontier.pop(0)
        reached.append(node["state"])
        
        if node["state"] == GOAL:
            return node
            
        for action in lay_hanh_dong(node["state"]):
            child_state = di_chuyen(node["state"], action)
            child = {"state": child_state, "parent": node, "action": action}
            
            in_reached = child_state in reached
            in_frontier = False
            for f in frontier:
                if f["state"] == child_state:
                    in_frontier = True
                    break
                    
            if not in_reached and not in_frontier:
                frontier.append(child)
                
    return None

# Main
if __name__ == '__main__':
    current_board = []
    for row in board:
        current_board.append(row[:])

    in_bang(current_board, "Trạng thái ban đầu:")
    in_bang(GOAL, "Trạng thái đích:")
    
    solution = breadth_first_search(current_board)
    
    if solution is not None:
        path = []
        curr = solution
        while curr["parent"] is not None:
            path.append((curr["action"], curr["state"]))
            curr = curr["parent"]
            
        path.reverse()
        
        for step, (action, state) in enumerate(path):
            print("Bước " + str(step + 1) + ": action='" + action + "'")
            in_bang(state)
        print("Đã hoàn thành!")
    else:
        print("Không tìm thấy giải pháp.")
```
