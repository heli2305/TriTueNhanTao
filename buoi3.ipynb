```python
import random

ROWS, COLS = 4, 4

# Môi trường
def init_environment():
    matrix = [[random.randint(0, 1) for _ in range(COLS)] for _ in range(ROWS)]
    robot = (random.randint(0, ROWS-1), random.randint(0, COLS-1))
    return matrix, robot


# Hành động
def get_actions(pos):
    r, c = pos
    actions = []
    if r > 0:        actions.append('U')
    if r < ROWS - 1: actions.append('D')
    if c > 0:        actions.append('L')
    if c < COLS - 1: actions.append('R')
    return actions

def execute_move(pos, action):
    r, c = pos
    if action == 'U': return (r-1, c)
    if action == 'D': return (r+1, c)
    if action == 'L': return (r, c-1)
    if action == 'R': return (r, c+1)
    return pos

# THUẬT TOÁN 
def rule_based_agent(matrix, robot):
    r, c = robot
    actions = get_actions(robot)

    if matrix[r][c] == 0:
        action = random.choice(actions)         
        robot = execute_move(robot, action)
        return matrix, robot, None, action

    if matrix[r][c] == 1:
        matrix[r][c] = 0                       
        action = random.choice(actions)         
        robot = execute_move(robot, action)
        return matrix, robot, 'S', action
    
# IN MA TRẬN
def print_matrix(matrix, robot):
    rr, rc = robot
    print('      ' + '   '.join(f'C{c}' for c in range(COLS)))
    print('    +' + '----+' * COLS)
    for r in range(ROWS):
        row = f'R{r}  |'
        for c in range(COLS):
            if (r, c) == (rr, rc):
                row += '[R] |'
            else:
                row += f' {matrix[r][c]}  |'
        print(row)
        print('    +' + '----+' * COLS)
    print()
```
