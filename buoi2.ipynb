```python
board = [[1, 2, 3], [4, 0, 6], [7, 5, 8]]

# In bảng
def in_bang(board, title=""):
    if title: print(title)
    for row in board:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()

# Vị trí ô trống
def tim_o_trong(board):
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c

# Di chuyển ô trống
def di_chuyen(board, action):
    new_board = [row[:] for row in board]
    r, c = tim_o_trong(new_board)
    moves = {'U': (r-1, c), 'D': (r+1, c), 'L': (r, c-1), 'R': (r, c+1)}
    nr, nc = moves[action]
    new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
    return new_board

# Các hướng di chuyển hợp lệ
def cam_bien(board):
    r, c = tim_o_trong(board)
    actions = []
    if r > 0: actions.append('U')
    if r < 2: actions.append('D')
    if c > 0: actions.append('L')
    if c < 2: actions.append('R')
    return actions

# Main
in_bang(board, "Trạng thái hiện tại:")
actions = cam_bien(board)
print(f"Các hành động có thể thực hiện: {actions}\n")
for action in actions:
    in_bang(di_chuyen(board, action), f"Sau {action}:")
```
